```python
# ==========================
# IMPORT LIBRARIES
# ==========================

import pandas as pd
import numpy as np
from faker import Faker

fake = Faker()

np.random.seed(42)

# ==========================
# LISTS
# ==========================

genders = ["Male", "Female"]

education_levels = [
    "High School",
    "Diploma",
    "Bachelor's",
    "Master's",
    "PhD"
]

departments = [
    "IT",
    "HR",
    "Finance",
    "Sales",
    "Marketing",
    "Operations",
    "Customer Support",
    "Administration",
    "Research",
    "Legal"
]

job_titles = [
    "Software Engineer",
    "Backend Developer",
    "Frontend Developer",
    "Full Stack Developer",
    "Data Analyst",
    "Data Scientist",
    "ML Engineer",
    "Data Engineer",
    "DevOps Engineer",
    "Cloud Engineer",
    "Cyber Security Analyst",
    "QA Engineer",
    "Database Administrator",
    "Business Analyst",
    "Product Manager",
    "Project Manager",
    "HR Executive",
    "HR Manager",
    "Finance Analyst",
    "Accountant",
    "Sales Executive",
    "Sales Manager",
    "Marketing Executive",
    "Marketing Manager",
    "Network Engineer"
]

cities = [
    "Delhi",
    "Noida",
    "Gurgaon",
    "Faridabad",
    "Ghaziabad",
    "Bangalore",
    "Hyderabad",
    "Pune",
    "Mumbai",
    "Chennai",
    "Kolkata",
    "Jaipur",
    "Ahmedabad",
    "Lucknow",
    "Indore"
]

work_modes = [
    "Office",
    "Hybrid",
    "Remote"
]

company_sizes = [
    "Startup",
    "Medium",
    "Large",
    "Enterprise"
]

marital_statuses = [
    "Single",
    "Married",
    "Divorced"
]

promotion_status = [
    "Yes",
    "No"
]

overtime = [
    "Yes",
    "No"
]

# ==========================
# BONUS DICTIONARIES
# ==========================

education_bonus = {
    "High School":0,
    "Diploma":5000,
    "Bachelor's":12000,
    "Master's":25000,
    "PhD":45000
}

job_bonus = {
    "Software Engineer":25000,
    "Backend Developer":27000,
    "Frontend Developer":22000,
    "Full Stack Developer":32000,
    "Data Analyst":18000,
    "Data Scientist":45000,
    "ML Engineer":48000,
    "Data Engineer":40000,
    "DevOps Engineer":35000,
    "Cloud Engineer":38000,
    "Cyber Security Analyst":36000,
    "QA Engineer":20000,
    "Database Administrator":28000,
    "Business Analyst":26000,
    "Product Manager":50000,
    "Project Manager":42000,
    "HR Executive":10000,
    "HR Manager":22000,
    "Finance Analyst":24000,
    "Accountant":15000,
    "Sales Executive":15000,
    "Sales Manager":30000,
    "Marketing Executive":17000,
    "Marketing Manager":32000,
    "Network Engineer":26000
}

company_bonus = {
    "Startup":0,
    "Medium":5000,
    "Large":12000,
    "Enterprise":22000
}

employees = []
```

```python
# ==========================
# GENERATE 5000 EMPLOYEES
# ==========================

for i in range(1, 5001):

    # Basic Details
    age = np.random.randint(21, 61)

    experience = np.random.randint(
        0,
        min(age - 20, 35) + 1
    )

    gender = np.random.choice(genders)

    education = np.random.choice(
        education_levels,
        p=[0.10, 0.20, 0.40, 0.25, 0.05]
    )

    department = np.random.choice(departments)

    job_title = np.random.choice(job_titles)

    performance = np.random.randint(1, 6)

    overtime_status = np.random.choice(
        overtime,
        p=[0.35, 0.65]
    )

    work_mode = np.random.choice(
        work_modes,
        p=[0.45, 0.35, 0.20]
    )

    city = np.random.choice(cities)

    company_size = np.random.choice(
        company_sizes,
        p=[0.20, 0.35, 0.30, 0.15]
    )

    marital_status = np.random.choice(
        marital_statuses,
        p=[0.45, 0.50, 0.05]
    )

    certification_count = np.random.randint(0, 11)

    skills_score = np.random.randint(40, 101)

    years_at_company = np.random.randint(
        0,
        min(experience, 20) + 1
    )

    promotion = np.random.choice(
        promotion_status,
        p=[0.25, 0.75]
    )

    annual_bonus = np.random.randint(
        20000,
        200001
    )

    # ==========================
    # SALARY CALCULATION
    # ==========================

    salary = 20000

    # Experience
    salary += experience * np.random.randint(2200, 3200)

    # Education
    salary += education_bonus[education]

    # Job Title
    salary += job_bonus[job_title]

    # Skills
    salary += skills_score * 180

    # Performance
    salary += performance * np.random.randint(2500, 4000)

    # Certifications
    salary += certification_count * 1200

    # Promotion
    if promotion == "Yes":
        salary += 12000

    # Overtime
    if overtime_status == "Yes":
        salary += 7000

    # Company Size
    salary += company_bonus[company_size]

    # Annual Bonus Impact
    salary += annual_bonus // 50

    # Random Noise
    salary += np.random.randint(-4000, 4001)

    salary = max(int(salary), 20000)

    # ==========================
    # STORE RECORD
    # ==========================

    employees.append({

        "Employee_ID": i,

        "Name": fake.name(),

        "Age": age,

        "Gender": gender,

        "Education": education,

        "Department": department,

        "Job_Title": job_title,

        "Experience": experience,

        "Performance_Rating": performance,

        "Overtime": overtime_status,

        "Work_Mode": work_mode,

        "City": city,

        "Company_Size": company_size,

        "Marital_Status": marital_status,

        "Certification_Count": certification_count,

        "Skills_Score": skills_score,

        "Years_At_Company": years_at_company,

        "Promotion_Last_5_Years": promotion,

        "Annual_Bonus": annual_bonus,

        "Monthly_Salary": salary

    })

# ==========================
# DATAFRAME
# ==========================

df = pd.DataFrame(employees)

print(df.shape)

df.head()
```

```python
# ====================================================
# ADD MISSING VALUES (Approx. 2%)
# ====================================================

np.random.seed(42)
columns_with_missing = [
    "Age",
    "Education",
    "Experience",
    "Performance_Rating",
    "Certification_Count",
    "Skills_Score",
    "Years_at_Company",
    "Department",
    "Work_Mode"
]

for col in columns_with_missing:
    missing_index = df.sample(frac=0.02, random_state=np.random.randint(1, 10000)).index
    df.loc[missing_index, col] = np.nan
    

# ====================================================
# ADD DUPLICATES (Approx. 1%)
# ====================================================
duplicate_rows = df.sample(frac=0.01, random_state=42)
df = pd.concat([df, duplicate_rows], ignore_index=True)

# ====================================================
# ADD OUTLIERS
# ====================================================
outlier_index = df.sample(frac=0.01, random_state=101).index
df.loc[outlier_index, "Monthly_Salary"] *= np.random.randint(
    2,
    4,
    len(outlier_index)
)

df.loc[outlier_index, "Annual_Bonus"] *= np.random.randint(
    2,
    5,
    len(outlier_index)
)

# ====================================================
# SHUFFLE DATASET
# ====================================================
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# ====================================================
# SAVE DATASET
# ====================================================
df.to_csv("../data/salary.csv", index=False)

# ====================================================
# VALIDATION
# ====================================================
print("="*50)
print("DATASET GENERATED SUCCESSFULLY")
print("="*50)

print("\nShape : ", df.shape)

print("\nMissing Values\n")
print(df.isnull().sum())

print("\nDuplicate Rows :", df.duplicated().sum())

print("\nData Types\n")
print(df.dtypes)

print("\nSummary Statistics\n")
print(df.describe())

print("\nTop 5 Rows\n")
display(df.head())
```
